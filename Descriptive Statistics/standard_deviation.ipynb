```python
# Data from the image
income_ranges = [(100, 150), (150, 200), (200, 250), (250, 300), (300, 350)]
frequencies = [1, 3, 6, 4, 1]

# Calculate midpoints
midpoints = [(low + high) / 2 for low, high in income_ranges]

# Calculate mean
total_frequency = sum(frequencies)
mean = sum(f * x for f, x in zip(frequencies, midpoints)) / total_frequency

# Calculate variance
variance = sum(f * ((x - mean) ** 2) for f, x in zip(frequencies, midpoints)) / total_frequency

# Standard deviation
std_deviation = variance ** 0.5

print(f"Mean: {mean}")
print(f"Variance: {variance}")
print(f"Standard Deviation: {std_deviation}")
```
