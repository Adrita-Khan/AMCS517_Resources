```python
import numpy as np

# Given data
X = np.array([10, 12, 16, 11, 15, 14, 20, 22], dtype=float)
Y = np.array([15, 18, 23, 14, 20, 17, 25, 28], dtype=float)

# 1) Regression of Y on X:  Y = a + bX
#   np.polyfit(X, Y, 1) returns [slope, intercept]
slope_Y_on_X, intercept_Y_on_X = np.polyfit(X, Y, 1)

print(f"Regression (Y on X): Y = {intercept_Y_on_X:.3f} + {slope_Y_on_X:.3f} X")

# 2) Regression of X on Y:  X = c + dY
#   np.polyfit(Y, X, 1) returns [slope, intercept]
slope_X_on_Y, intercept_X_on_Y = np.polyfit(Y, X, 1)

print(f"Regression (X on Y): X = {intercept_X_on_Y:.3f} + {slope_X_on_Y:.3f} Y")
```

```python
import numpy as np

# Given data
# X: No. of families (in lakhs)
# Y: Sale of automobiles (in '000)
X = np.array([70, 75, 80, 60, 90], dtype=float)
Y = np.array([25.2, 28.6, 30.2, 22.3, 35.4], dtype=float)

# Fit linear regression Y on X: Y = a + bX
slope, intercept = np.polyfit(X, Y, 1)

print(f"Regression Equation (Y on X): Y = {intercept:.3f} + {slope:.3f} X")

# Predict Y when X = 100 (for Belagavi in 2020)
X_new = 100
Y_pred = intercept + slope * X_new

print(f"Estimated sale of automobiles for Belagavi (100 lakh families): {Y_pred:.2f} ('000 units)")
```

```python
import math

# 1. Data
X = [42, 36, 55, 58, 35, 65, 60, 50, 48, 51]  # Ages
Y = [98, 93, 110, 85, 105, 108, 82, 102, 118, 99]  # Blood Pressures

# 2. Compute means
n = len(X)
mean_x = sum(X) / n
mean_y = sum(Y) / n

# 3a. Compute slope and intercept for Y on X
#     slope b_yx = Σ((x - x̄)(y - ȳ)) / Σ((x - x̄)^2)
#     intercept a_yx = ȳ - b_yx * x̄
num_yx = 0.0
den_yx = 0.0
for i in range(n):
    dx = X[i] - mean_x
    dy = Y[i] - mean_y
    num_yx += dx * dy
    den_yx += dx**2

b_yx = num_yx / den_yx
a_yx = mean_y - b_yx * mean_x

# 3b. Compute slope and intercept for X on Y
#     slope b_xy = Σ((x - x̄)(y - ȳ)) / Σ((y - ȳ)^2)
#     intercept a_xy = x̄ - b_xy * ȳ
num_xy = 0.0
den_xy = 0.0
for i in range(n):
    dx = X[i] - mean_x
    dy = Y[i] - mean_y
    num_xy += dx * dy
    den_xy += dy**2

b_xy = num_xy / den_xy
a_xy = mean_x - b_xy * mean_y

# 4. Compute correlation coefficient r from the slopes
#    For simple linear regression, b_yx * b_xy = r^2  (assuming a positive correlation)
#    We can check the sign by looking at the sign of num_yx (the covariance).
cov_xy = num_yx / n
if cov_xy >= 0:
    r = math.sqrt(b_yx * b_xy)
else:
    r = -math.sqrt(b_yx * b_xy)

# 5. Estimate blood pressure when Age = 45 using Y = a_yx + b_yx * X
age_to_predict = 45
predicted_bp = a_yx + b_yx * age_to_predict

# Print results
print(f"Regression of Y on X:   Y = {a_yx:.3f} + {b_yx:.3f}X")
print(f"Regression of X on Y:   X = {a_xy:.3f} + {b_xy:.3f}Y")
print(f"Correlation coefficient r = {r:.3f}")
print(f"Estimated BP at age 45 = {predicted_bp:.2f}")
```
