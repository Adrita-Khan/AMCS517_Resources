```python
import numpy as np

# Advertisement Expenses and Sales Volume
advertisement_exp = [11, 13, 14, 16, 16, 15, 15, 14, 13, 13]
sales_volume = [50, 50, 55, 60, 65, 65, 65, 60, 60, 50]

# Calculate the correlation coefficient
correlation_coefficient = np.corrcoef(advertisement_exp, sales_volume)[0, 1]

print("Karl Pearson's Correlation Coefficient:", round(correlation_coefficient, 4))
```

```python
import numpy as np

# Given data
ages = [15, 16, 17, 18, 19, 20]
number_of_students = [250, 200, 150, 120, 100, 80]
regular_players = [200, 150, 90, 48, 30, 12]

# Since correlation should be between age and proportion of regular players, calculate the ratio
playing_ratio = [rp/ns for rp, ns in zip(regular_players, number_of_students)]

# Calculate correlation coefficient between age and playing ratio
correlation_coefficient = np.corrcoef(ages, playing_ratio)[0, 1]

print("Karl Pearson's Correlation Coefficient (Age vs. Playing Ratio):", round(correlation_coefficient, 4))
```
